```
# 1. Clase Persona
# Encapsulamiento con atributos privados

class Persona:
    def __init__(self, edad):
        self.__edad = edad

    # Getter
    def obtener_edad(self):
        return self.__edad

    # Setter
    def cambiar_edad(self, nueva_edad):
        if nueva_edad > 0:
            self.__edad = nueva_edad
        else:
            print("La edad debe ser mayor a 0")


p1 = Persona(18)

print(p1.obtener_edad())

p1.cambiar_edad(25)

print(p1.obtener_edad())
```

```
# 3. Validar saldo negativo

class Cuenta:
    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, monto):
        if monto > 0:
            self.__saldo += monto
        else:
            print("No se permiten depósitos negativos")

    def mostrar_saldo(self):
        print("Saldo:", self.__saldo)


c1 = Cuenta(1000)

c1.depositar(500)

c1.mostrar_saldo()
```

```
# 4. Retiro de dinero

class Cuenta:
    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, monto):
        if monto > 0:
            self.__saldo += monto

    def retirar(self, monto):
        if monto > self.__saldo:
            print("Fondos insuficientes")
        else:
            self.__saldo -= monto

    def mostrar_saldo(self):
        print("Saldo:", self.__saldo)


c1 = Cuenta(1000)

c1.retirar(300)

c1.mostrar_saldo()
```

```
# 5. Estudiante

class Estudiante:
    def __init__(self):
        self.__nota = 0

    def asignar_nota(self, nota):
        if 0 <= nota <= 5:
            self.__nota = nota
        else:
            print("Nota inválida")

    def mostrar_nota(self):
        print("Nota:", self.__nota)


e1 = Estudiante()

e1.asignar_nota(4.5)

e1.mostrar_nota()
```

```
# 6. Producto

class Producto:
    def __init__(self, precio):
        self.__precio = precio

    def modificar_precio(self, nuevo_precio):
        if nuevo_precio > 0:
            self.__precio = nuevo_precio
        else:
            print("El precio debe ser positivo")

    def mostrar_precio(self):
        print("Precio:", self.__precio)


p1 = Producto(2000)

p1.modificar_precio(3500)

p1.mostrar_precio()
```

```
# 7. Temperatura

class Temperatura:
    def __init__(self, celsius):
        self.__celsius = celsius

    def obtener_temperatura(self):
        return self.__celsius

    def cambiar_temperatura(self, nueva_temp):
        if -100 <= nueva_temp <= 100:
            self.__celsius = nueva_temp
        else:
            print("Temperatura fuera de rango")


t1 = Temperatura(25)

t1.cambiar_temperatura(30)

print(t1.obtener_temperatura())
```

```
# 8. Usuario con contraseña

class Usuario:
    def __init__(self, password):
        self.__password = password

    def cambiar_password(self, nueva_password):
        if len(nueva_password) >= 6:
            self.__password = nueva_password
        else:
            print("La contraseña debe tener mínimo 6 caracteres")

    def mostrar_password(self):
        print(self.__password)


u1 = Usuario("123456")

u1.cambiar_password("abcdef")

u1.mostrar_password()
```

```
# 9. Vehículo

class Vehiculo:
    def __init__(self):
        self.__velocidad = 0

    def acelerar(self, aumento):
        if aumento >= 0:
            self.__velocidad += aumento
        else:
            print("La velocidad no puede ser negativa")

    def mostrar_velocidad(self):
        print("Velocidad:", self.__velocidad)


v1 = Vehiculo()

v1.acelerar(60)

v1.mostrar_velocidad()
```

```
# 10. CuentaBancaria completa

class CuentaBancaria:
    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, monto):
        if monto > 0:
            self.__saldo += monto
        else:
            print("No se permiten depósitos negativos")

    def retirar(self, monto):
        if monto <= 0:
            print("El retiro debe ser positivo")
        elif monto > self.__saldo:
            print("Saldo insuficiente")
        else:
            self.__saldo -= monto

    def obtener_saldo(self):
        return self.__saldo


cuenta = CuentaBancaria(1000)

cuenta.depositar(500)

cuenta.retirar(200)

print("Saldo actual:", cuenta.obtener_saldo())
```
